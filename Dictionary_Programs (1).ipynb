```python
dic={101:"Mango",102:"Banana",103:"Grapes"}
print("Keys")
for key in dic:
    print(key)
print("\nValues:")
for value in dic.values():
    print(value)
print("\nKeys and values:")
for key,value in dic.items():
 print(key,value)
```
