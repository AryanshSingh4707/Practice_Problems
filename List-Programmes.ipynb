```python
#Reverse
numbers = [10,20,30,40]
print("original List",numbers)
numbers.reverse()
print("Reversed List",numbers)
```

```python
#Append
numbers = [10,20,30,40]
print("original List",numbers)
numbers.append(60)
print("After appending",numbers)
```

```python
#Remove
numbers = [10,20,30,40]
print("original List",numbers)
numbers.remove(20)
print("List after removal",numbers)
```

```python
#Pop
numbers = [10,20,30,40]
print("original List",numbers)
numbers.pop(1)
print("List after removal",numbers)
```
